Challange 2 -  Update Store Database

```python
#Part_1 - Create New List

product = ['T-shirt','Blouse','Shirt']
price = [59.99, 299.99, 80.00]
```

```python
#Part_2 - Register New Product


new_products = input("Do you want upload a new product? ")

if new_products.lower() == "yes":
  new_product = input("New product name: ")
  new_price = float(input("New product price: "))
  print("Database Updated")

  product.append(new_product)
  price.append(new_price)
  for i,j in zip (product,price):
    print(f"The product {i} price is US$ {j:.2f}")
```

```python
#Parte_3 - Remove Last Product

remove = len(product)

if remove > 0:
  product.pop()
  price.pop()
  print("Last item removed")
  for i,j in zip (product,price):
    print(f"The product {i} price is US$ {j:.2f}")
else:
  print("Stock Empty")
```

```python
#Part_4 - Find and Remove Product

item = input("Do you like to remove any item? ")

if item in product:
    indice = product.index(item)
    product.pop(indice)
    price.pop(indice)
    print("Product Removed")
    for i in range(len(product)):
        print(f"The product {product[i]} price is US${price[i]:.2f}")
else:
    print("Product not found")
```
